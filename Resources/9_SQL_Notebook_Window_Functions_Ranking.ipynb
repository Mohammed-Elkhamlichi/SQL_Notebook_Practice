# **Cohort Year Impact On Customer Growth**

```python
%%sql
WITH cohort_year AS(
    SELECT
        customerkey,
        EXTRACT(YEAR FROM MIN(orderdate) OVER(PARTITION BY customerkey)) AS cohort_year,
        EXTRACT(YEAR FROM orderdate) AS purchase_year
    FROM
        sales
)
SELECT DISTINCT
    cy.cohort_year,
    cy.purchase_year,
    COUNT(customerkey) OVER(PARTITION BY purchase_year, cohort_year) AS unique_customers
FROM cohort_year cy

ORDER BY cohort_year, purchase_year
```
